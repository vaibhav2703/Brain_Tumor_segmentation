```python
from monai.networks.nets import UNet
from monai.networks.layers import Norm
from monai.losses import DiceLoss, DiceCELoss
```

```python
import torch
import monai
```

```python
from preprocess import prepare
from utilities import train
```

```python
data_dir = 'D:\Brain_Data'
model_dir = 'D:\Brain_Tumor_Segmentation\Model'
data_in = prepare(data_dir)
```

```python
device = torch.device("cpu")
```

```python
model = UNet(
    spatial_dims=3,  # Specify the spatial dimensions
    in_channels=4,
    out_channels=3,
    channels=(16, 32, 64, 128, 256),
    strides=(2, 2, 2, 2),
    num_res_units=2,
    norm=Norm.BATCH,
).to(device)

# model = UNet(
#     spatial_dims=3,
#     in_channels=1,
#     out_channels=1,
#     channels=(240, 480, 960, 1920, 3840),  # Adjusted channel sizes
#     strides=(2, 2, 2, 2),
#     num_res_units=2,
#     norm=Norm.BATCH,
# ).to(device)

```

```python
sample_input = torch.randn(4,4 , 64, 64, 64).to(device)
```

```python
output = model(sample_input)

# Print out the shapes of tensors at different points in the model
print("Input shape:", sample_input.shape)
print("Output shape:", output.shape)
```

```python
output.shape[1] == 2, f"Expected utput channels but got {output.shape[1]}"
```

```python
loss_fuction=DiceLoss(to_onehot_y = True, sigmoid=True, squared_pred=True)
```

```python
optimizer = torch.optim.Adam(model.parameters(), 1e-5, weight_decay=1e-5, amsgrad=True)
```

```python
if __name__ == '__main__':
    train(model, data_in, loss_fuction, optimizer, 5, model_dir)
```
